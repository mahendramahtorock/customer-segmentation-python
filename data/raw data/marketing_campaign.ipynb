# EDA 5: Product Spending Analysis
Business Question

Which product categories do customers spend the most money on?

1-MntWines
2-MntFruits
3-MntMeatProducts
4-MntFishProducts
5-MntSweetProducts
6-MntGoldProds

# Calculating total spending for each product

```python
product_spending = df[[ "MntWines",
        "MntFruits",
        "MntMeatProducts",
        "MntFishProducts",
        "MntSweetProducts",
        "MntGoldProds"]].sum()

product_spending
```

```python
product_spending.index=[
    "Wine",
    "Fruits",
    "Meat",
    "Fish",
    "Sweets",
    "Gold"
]
```

```python
product_spending = df[
    [
        "MntWines",
        "MntFruits",
        "MntMeatProducts",
        "MntFishProducts",
        "MntSweetProducts",
        "MntGoldProds"
    ]
].sum()


    
product_share=(product_spending/product_spending.sum()*100).sort_values(ascending=False)
product_share
```

```python
product_spending
```

```python
plt.figure(figsize=(10,6))

product_spending.sort_values(ascending = False).plot(
     kind="bar"
 )

plt.title("Total Spending by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Spending")

plt.xticks(rotation = 45)
plt.show()
```

# Business Insights
Wine has the highest customer spending among all product categories.
Meat products are the second-highest spending category.
Fruits, fish, and sweets have relatively lower spending.
Gold products also contribute a good amount of customer spending.
The company should focus more on Wine and Meat products while finding ways to increase sales of lower-performing categories.

# EDA 6: Which channel do customers use the most?
 Business Question

Do customers prefer buying through the website, catalog, or physical store?

```python
purchase_channels= df[[
    "NumWebPurchases",
        "NumCatalogPurchases",
        "NumStorePurchases"
]].sum()

purchase_channels
```

```python
purchase_channels.index= [
    "Web",
    "Catalog",
    "Store"
]
```

```python
plt.figure(figsize=(8,5))

purchase_channels.sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Total Purchases by Channel")
plt.xlabel("Purchase Channel")
plt.ylabel("Number of Purchases")

plt.xticks(rotation=0)

plt.show()
```

# Business Insights
Store purchases are the highest, showing that customers prefer buying from physical stores.
Web purchases are the second most-used channel.
Catalog purchases are the lowest among the three channels.
The company can focus on improving its online and catalog experience to increase purchases through these channels.
Since stores are the strongest channel, the company should continue supporting its physical-store customers.

# EDA 8 — Does Income Affect Spending?
🎯 Business Question

Do customers with higher income tend to spend more?

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data= df,
    x= "Income",
    y ="Total_Spending"
)
plt.title("Income vs Total Spending")
plt.xlabel("Income")
plt.ylabel("Total Spending")

plt.show()
```

# Business Insights – Income vs Total Spending
1-Most customers have an income between 20,000 and 80,000 and their spending is generally lower than high-income customers.

2-Customers with income above 80,000 show higher spending potential.

3-A few customers have income above 100,000, making them potential high-value customers.

4-The company can target high-income customers with premium products and personalized offers.

5-For customers in the 20,000–80,000 range, the company can use discounts and bundle offers to increase spending.

6-This can help the company increase revenue by using different offers for different income groups.


# EDA 9 — Does Age Affect Spending?
Business Question

Do older customers spend more than younger customers?

This helps the business understand whether age should be considered when creating marketing offers.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data = df,
    x="Age",
    y = "Total_Spending"
)

plt.title("Age vs Total Spending")
plt.xlabel("Age")
plt.ylabel("Total Spending")

plt.show()
```

# Business Insights – Age vs Total Spending
Customers aged around 40–60 years show stronger spending levels.
Customers below 30 years generally have lower spending.
Spending varies across age groups, so age alone does not fully explain customer spending.
The company can target the 40–60 age group with premium products and personalized offers.
For younger customers, discounts and attractive bundles could help increase spending.
The company should also remove unrealistic age values before making final business decisions.
